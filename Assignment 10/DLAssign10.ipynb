```python

# ============================================================
# DEEP LEARNING PRACTICAL NO. 10
# Audio Processing: Speech-to-Text and Text-to-Speech
# ============================================================

# CELL 1: Install required libraries
!pip -q install openai-whisper gTTS jiwer
!apt-get -qq update
!apt-get -qq install -y ffmpeg

print("Required libraries installed successfully!")


# CELL 2: Import libraries
import os
import time
import whisper

from gtts import gTTS
from jiwer import wer
from google.colab import files
from IPython.display import Audio, display, Markdown

print("Libraries imported successfully!")


# CELL 3: Load Whisper deep learning model
model = whisper.load_model("base")

print("Whisper model loaded successfully!")


# CELL 4: Upload an audio file
print("Upload an audio file (WAV, MP3, or M4A):")

uploaded = files.upload()

audio_file = next(iter(uploaded.keys()))

print("Uploaded file:", audio_file)


# CELL 5: Speech-to-Text
start_time = time.time()

result = model.transcribe(audio_file)

stt_time = time.time() - start_time

recognized_text = result["text"].strip()
detected_language = result.get("language", "unknown")

print("\n========== SPEECH-TO-TEXT RESULTS ==========")
print("Input audio:", audio_file)
print("Recognized text:", recognized_text)
print("Detected language:", detected_language)
print(f"Processing time: {stt_time:.2f} seconds")

print("\nOriginal audio:")
display(Audio(filename=audio_file))


# CELL 6: Evaluate Speech-to-Text using WER
reference_text = input(
    "Enter the correct transcript of the uploaded audio: "
).strip()

if reference_text and recognized_text:
    error_rate = wer(
        reference_text.lower(),
        recognized_text.lower()
    )

    print("\n========== STT EVALUATION ==========")
    print("Reference transcript:", reference_text)
    print("Recognized transcript:", recognized_text)
    print(f"Word Error Rate (WER): {error_rate:.4f}")
    print(f"Word Error Rate: {error_rate * 100:.2f}%")

    if error_rate == 0:
        print("Result: Exact word match.")
    else:
        print("Result: Transcription contains word errors.")

    if error_rate <= 1:
        print(f"Approximate word accuracy: {(1-error_rate)*100:.2f}%")
    else:
        print("Word accuracy: Not reported because WER exceeds 100%.")
else:
    error_rate = None
    print("Reference or recognized text is empty. Evaluation skipped.")


# CELL 7: Text-to-Speech
print("\n========== TEXT-TO-SPEECH ==========")

if recognized_text:
    tts_start_time = time.time()

    tts = gTTS(
        text=recognized_text,
        lang=detected_language
        if detected_language in ["en", "hi", "fr", "es", "de", "it", "pt"]
        else "en"
    )

    output_file = "/content/generated_speech.mp3"
    tts.save(output_file)

    tts_time = time.time() - tts_start_time

    print("Speech generated successfully!")
    print(f"TTS generation time: {tts_time:.2f} seconds")

    print("\nGenerated speech:")
    display(Audio(filename=output_file))

    files.download(output_file)
else:
    tts_time = 0
    print("No recognized text available for speech generation.")


# CELL 8: Display recognized text
print("\n========== RECOGNIZED TEXT ==========")

display(Markdown("### Speech-to-Text Output"))
display(Markdown(recognized_text if recognized_text else "No text recognized."))


# CELL 9: Final performance evaluation
print("\n" + "=" * 50)
print("FINAL PROJECT RESULTS")
print("=" * 50)

print("Input audio file:", audio_file)
print("Detected language:", detected_language)
print("Recognized text:", recognized_text)
print(f"STT processing time: {stt_time:.2f} seconds")
print(f"TTS generation time: {tts_time:.2f} seconds")

if error_rate is not None:
    print(f"Word Error Rate: {error_rate * 100:.2f}%")

    if error_rate <= 1:
        print(f"Approximate word accuracy: {(1-error_rate)*100:.2f}%")

if os.path.exists("/content/generated_speech.mp3"):
    print("TTS output: Generated successfully")
    print("Output format: MP3")
else:
    print("TTS output: Not generated")

print("=" * 50)
print("Program completed successfully!")
```
